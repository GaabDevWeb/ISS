```python
import pandas as pd
import sqlite3
```

```python
# 1. Caminhos do Excel e do Banco de Dados
caminho_excel = r'D:\Temp\excel_Cliente.xlsx'
nome_banco = r'D:\SQLLite\Databases\Databases.db' # O arquivo será criado na mesma pasta do script
```

```python
# 2. Conecta ao banco (se não existir, ele cria na hora)
conexao = sqlite3.connect(nome_banco)
cursor = conexao.cursor()
```

```python
# 3. Cria a tabela (ajuste os nomes das colunas conforme seu Excel)
cursor.execute('''
    CREATE TABLE IF NOT EXISTS clientes (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        nome TEXT,
        email TEXT,
        telefone TEXT
    )
''')
```

```python
# 4. Carrega o Excel
df = pd.read_excel(caminho_excel)
```

```python
# 5. Percorre linha a linha e grava no SQLite
print("Iniciando a gravação no banco...")

for indice, linha in df.iterrows():
    # Pegando os dados das colunas do Excel (Ajuste os nomes entre [''])
    nome = linha['Nome']
    email = linha['Email']
    tel = linha['Telefone']
    
    # Insere no banco de dados
    cursor.execute("INSERT INTO clientes (nome, email, telefone) VALUES (?, ?, ?)", (nome, email, tel))

# 6. Salva as alterações e fecha a conexão
conexao.commit()
conexao.close()

print(f"Sucesso! {len(df)} linhas foram gravadas no arquivo '{nome_banco}'.")
```
